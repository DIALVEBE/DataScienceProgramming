| Estructura | Sintaxis | Mutable | Ordenada | Acceso |
|---|---|---|---|---|
| Lista | `[1, 2, 3]` | Si | Si | Por indice, desde 0 |
| Tupla | `(1, 2)` | No | Si | Por indice |
| Diccionario | `{'k': v}` | Si | Si (3.7+) | Por clave |
| Conjunto | `{1, 2}` | Si | No | Sin indice, pertenencia O(1) |

| Pieza | Nota rapida |
|---|---|
| `def f(x): return y` | Sin `return` devuelve `None` |

| Comprension | `[f(x) for x in xs if cond]`; tambien de dict y de set |
| `enumerate` / `zip` | Indice y valor / recorrer en paralelo |
| Desempaquetado | `a, b = b, a`; `primero, *resto = lista` |
| `dict.get(k, default)` | Evita el `KeyError` |
| `collections` | `defaultdict`, `Counter`, `namedtuple` |
| Truthiness | `[]`, `{}`, `''`, `0`, `None` son falsos |

Dos cosas que la clase normal no alcanza a ver y que aqui vas a usar:

| Pieza | Para que |

```python
# --- verificacion del reto 4 ---
plan, sobrante = plan_reposicion(CATALOGO, presupuesto=500000, stock_minimo=50)

assert isinstance(plan, list) and isinstance(sobrante, int), 'Devuelve (lista, int)'
assert all(isinstance(t, tuple) and len(t) == 3 for t in plan), 'Tuplas de 3 elementos'

nombres = [p for p,_,_ in plan]
assert 'Arroz' not in nombres, 'Arroz tiene 200 unidades: esta por encima del minimo'
assert 'Sal' not in nombres, 'Sal tiene 90: tampoco necesita'

costo_total = sum(c for _,_,c in plan)
assert costo_total + sobrante == 500000, 'Lo gastado mas lo sobrante es el presupuesto'
assert sobrante >= 0, 'El sobrante no puede ser negativo'

for prod, unid, costo in plan:
    precio = next(p for lst in CATALOGO.values() for (n,p,_) in lst if n == prod)
    stock  = next(u for lst in CATALOGO.values() for (n,_,u) in lst if n == prod)
    assert unid == 50 - stock, f'{prod}: se repone exactamente hasta el minimo'
    assert costo == unid * precio, f'{prod}: costo = unidades * precio'

costos = [c for _,_,c in plan]
assert costos == sorted(costos), 'Se aprueban de menor a mayor costo del faltante'

vacio, sob0 = plan_reposicion(CATALOGO, 0, 50)
assert vacio == [] and sob0 == 0, 'Sin presupuesto no hay plan'

# con presupuesto justo para el mas barato y nada mas
chico = {'A': [('Caro', 1000, 0), ('Barato', 10, 0)]}
p2, s2 = plan_reposicion(chico, presupuesto=100, stock_minimo=5)
assert [n for n,_,_ in p2] == ['Barato'], 'Solo alcanza el barato'
assert s2 == 100 - 50, 'Barato: 5 unidades a 10 = 50'

# el caro no cabe pero no debe bloquear al que viene despues
salto = {'A': [('Enorme', 100000, 0), ('Chico', 5, 0)]}
p3, _ = plan_reposicion(salto, presupuesto=1000, stock_minimo=10)
assert [n for n,_,_ in p3] == ['Chico'], \
    'Si uno no cabe se salta, pero se siguen evaluando los demas'

print('Reto 4 resuelto.')
print(f'\nPlan con 500.000 de presupuesto: {len(plan)} productos, '
      f'sobran {sobrante:,}')
for prod, unid, costo in plan:
    print(f'  {prod:14} {unid:4} unidades  ${costo:>9,}')
```
