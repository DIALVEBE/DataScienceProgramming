## Parte 6 - Listas

Se escriben entre corchetes `[ ]`. Se accede por posicion, **empezando en 0**.

```
productos = ['Cafe', 'Pan', 'Leche']
                0      1       2
```

````python
productos = ['Cafe', 'Pan', 'Leche', 'Queso']
precios = [3500, 1200, 4800, 12000]

print(productos)
print(f'La lista tiene {len(productos)} elementos')

print('Primero:', productos[0])
print('Tercero:', productos[2])
print('Ultimo:', productos[-1])
````

### 6.1 Las listas se pueden modificar

| Operacion | Que hace |
|---|---|
| `lista.append(x)` | agrega x al final |
| `lista.remove(x)` | borra la primera aparicion de x |
| `lista[i] = x` | reemplaza la posicion i |
| `sum()` `max()` `min()` `sorted()` | operan sobre toda la lista |

````python
precios = [3500, 1200, 4800, 12000]

precios.append(2000)
print('Despues de append:', precios)

precios[0] = 4000
print('Despues de cambiar:', precios)

print('Total:', sum(precios))
print('El mas caro:', max(precios))
print('Promedio:', sum(precios) / len(precios))
````

## Parte 8 - Diccionarios

Se escriben entre llaves `{ }`, en pares **clave: valor**. Se busca por clave, no por posicion.

```
precios = {'Cafe': 3500, 'Pan': 1200}
            clave  valor  clave valor
```

````python
precios = {
    'Cafe': 3500,
    'Pan': 1200,
    'Leche': 4800,
    'Queso': 12000
}

print('El cafe cuesta:', precios['Cafe'])

precios['Huevos'] = 900
print(precios)

for producto, precio in precios.items():
    print(f'{producto:10} ${precio:>7,}')

print('\nSolo las claves:', list(precios.keys()))
print('Solo los valores:', list(precios.values()))
````

### 8.1 Cuando la clave no existe

````python
print(precios['Arroz'])

print(precios.get('Arroz', 0))

if 'Arroz' in precios:
    print('Si hay arroz')
else:
    print('No tenemos arroz en el catalogo')
````

### 8.2 Diccionarios anidados

````python
supermercado = {
    'Frutas':   [('Manzana', 4500), ('Banano', 2800)],
    'Lacteos':  [('Leche', 4800), ('Queso', 12000), ('Yogur', 3200)],
    'Vegetales':[('Tomate', 3900), ('Cebolla', 2500)]
}

for categoria, lista_productos in supermercado.items():
    print(f'\n{categoria}:')
    for nombre_producto, precio in lista_productos:
        print(f'   {nombre_producto:10} ${precio:,}')

print('\nPrimer lacteo:', supermercado['Lacteos'][0])
print('Nombre del primer lacteo:', supermercado['Lacteos'][0][0])
````

## Parte 9 - Funciones

```python
def nombre_de_la_funcion(parametro1, parametro2):
    return resultado
```

Definir una funcion no la ejecuta. Solo corre cuando la llamas por su nombre.

````python
def saludar(nombre):
    mensaje = f'Hola, {nombre}! Bienvenido al curso.'
    return mensaje

print(saludar('Ana'))
print(saludar('Carlos'))
````

### 9.1 return no es lo mismo que print

- `print()` muestra en pantalla y el valor se pierde.
- `return` devuelve el valor al programa para poder usarlo.

````python
def sumar_mal(a, b):
    print(a + b)

def sumar_bien(a, b):
    return a + b

resultado_a = sumar_mal(2, 3)
resultado_b = sumar_bien(2, 3)

print('Lo que devolvio sumar_mal:', resultado_a)
print('Lo que devolvio sumar_bien:', resultado_b)

print('El doble:', sumar_bien(2, 3) * 2)
print('El doble del malo:', sumar_mal(2, 3) * 2)
````

### 9.2 Las dos funciones que pide la actividad

````python
def total_productos(categoria):
    return len(supermercado[categoria])


def precio_promedio(categoria):
    lista_productos = supermercado[categoria]

    if len(lista_productos) == 0:
        return 0

    solo_precios = []
    for nombre_producto, precio in lista_productos:
        solo_precios.append(precio)

    return sum(solo_precios) / len(solo_precios)


for categoria in supermercado:
    cuantos = total_productos(categoria)
    promedio = precio_promedio(categoria)
    print(f'{categoria:10} {cuantos} productos | promedio ${promedio:,.0f}')
````

## Parte 10 - DEMO: juntamos todo

Diccionario + listas + tuplas + for + if + funciones en un solo programa.

````python
cafeteria = {
    'Bebidas':   [('Tinto', 2000), ('Capuchino', 6500), ('Jugo', 5000)],
    'Panaderia': [('Croissant', 4500), ('Pan de queso', 2800)],
    'Postres':   [('Brownie', 7500), ('Cheesecake', 11000)]
}

CATEGORIA_EN_DESCUENTO = 'Panaderia'
DESCUENTO = 0.10
UMBRAL_CARO = 5000


def aplicar_descuento(precio, porcentaje):
    return precio * (1 - porcentaje)


def clasificar(precio):
    if precio > UMBRAL_CARO:
        return 'caro'
    else:
        return 'barato'


for categoria, productos in cafeteria.items():
    print(f'\n{categoria}')
    print('-' * 40)

    for nombre_producto, precio in productos:

        if categoria == CATEGORIA_EN_DESCUENTO:
            precio_final = aplicar_descuento(precio, DESCUENTO)
            marca = f'(-{DESCUENTO:.0%})'
        else:
            precio_final = precio
            marca = ''

        etiqueta = clasificar(precio_final)
        print(f'  {nombre_producto:14} ${precio_final:>8,.0f}  {etiqueta:7} {marca}')
````

## Resumen

| Concepto | Ejemplo | Para que |
|---|---|---|
| Mostrar | `print('Hola')` | ver que esta pasando |
| Variable | `edad = 30` | guardar un valor con nombre |
| Tipos | `int`, `float`, `str`, `bool` | determinan que operaciones valen |
| Ver tipo | `type(edad)` | depurar |
| f-string | `f'{nombre} tiene {edad}'` | armar texto con variables |
| Decidir | `if / elif / else` | elegir camino |
| Repetir | `for x in coleccion:` | recorrer datos |
| Repetir mientras | `while condicion:` | repetir sin saber cuantas veces |
| Lista | `['Cafe', 'Pan']` | muchos valores, ordenados, modificables |
| Tupla | `('Cafe', 3500)` | valores que van juntos y no cambian |
| Diccionario | `{'Cafe': 3500}` | buscar por nombre, no por posicion |
| Funcion | `def f(x): return x*2` | empaquetar trabajo reutilizable |

### Los cinco errores que viste hoy

| Error | Que significa |
|---|---|
| `TypeError` | mezclaste tipos incompatibles |
| `IndentationError` | falta o sobra sangria |
| `IndexError` | pediste una posicion que no existe |
| `KeyError` | pediste una clave que no esta en el diccionario |
| `NoneType` en una operacion | usaste el resultado de una funcion sin `return` |
