## Parte 2 - El array de NumPy

Un array se parece a una lista, pero por dentro es otra cosa:

| Razon | Que significa |
|---|---|
| **Tipos homogeneos** | todos los elementos son del mismo tipo |
| **Memoria contigua** | los datos van juntos en un solo bloque de memoria |

`import numpy as np` trae la libreria y le pone el apodo `np`.

````python
import numpy as np

precios = np.array([100, 200, 300, 400])

print(precios)
print('Tipo del objeto:', type(precios))
print('Tipo de los datos:', precios.dtype)
print('Forma:', precios.shape)
print('Dimensiones:', precios.ndim)
print('Cuantos elementos:', precios.size)
````

### 3.2 De verdad es mas rapido? Midamoslo

````python
import time

N = 1000000

lista_python = list(range(N))
array_numpy = np.arange(N)

inicio = time.time()
resultado_lista = []
for x in lista_python:
    resultado_lista.append(x * 1.19)
tiempo_lista = time.time() - inicio

inicio = time.time()
resultado_array = array_numpy * 1.19
tiempo_numpy = time.time() - inicio

print(f'Bucle de Python : {tiempo_lista:.4f} segundos')
print(f'NumPy vectorizado: {tiempo_numpy:.4f} segundos')
print(f'NumPy fue aprox. {tiempo_lista / max(tiempo_numpy, 1e-9):.0f} veces mas rapido')
````

## Parte 4 - Agregacion: resumir muchos numeros en uno

| Metodo | Que devuelve |
|---|---|
| `.mean()` | el promedio |
| `.sum()` | la suma de todos |
| `.max()` / `.min()` | el mayor / el menor |
| `.std()` | la desviacion estandar |
| `.argmax()` | la posicion del mayor |
| `np.median(a)` | la mediana |

````python
notas = np.array([4.5, 3.2, 2.8, 4.9, 3.7, 2.5, 4.0, 3.9, 1.8, 4.6])

print('Cuantas notas :', notas.size)
print('Promedio      :', notas.mean().round(2))
print('Mediana       :', np.median(notas))
print('Maxima        :', notas.max())
print('Minima        :', notas.min())
print('Suma          :', notas.sum())
print('Desviacion    :', notas.std().round(2))

posicion = notas.argmax()
print(f'La nota mas alta esta en la posicion {posicion} y vale {notas[posicion]}')
````

## Parte 5 - Filtrado booleano

1. `notas >= 3.0` produce una **mascara**: un array de `True` y `False`.
2. `notas[notas >= 3.0]` se queda solo con las posiciones en `True`.

```
notas       = [4.5, 3.2, 2.8, 4.9]
notas >= 3  = [True, True, False, True]
notas[...]  = [4.5, 3.2,        4.9]
```

````python
mascara = notas >= 3.0
print('Mascara:', mascara)
print('Tipo de la mascara:', mascara.dtype)

aprobados = notas[mascara]
print('Aprobados:', aprobados)

print('Aprobados (una linea):', notas[notas >= 3.0])

print('Cuantos aprobaron:', mascara.sum())
print(f'Tasa de aprobacion: {mascara.mean():.0%}')
````

### 5.1 Combinar condiciones: `&` y `|`, nunca `and` ni `or`

| En un `if` de Python | En un array de NumPy |
|---|---|
| `and` | `&` |
| `or` | `\|` |
| `not` | `~` |

````python
notables = notas[(notas >= 3.0) and (notas <= 4.5)]
print('Entre 3.0 y 4.5:', notables)

extremos = notas[(notas < 2.0) | (notas >= 4.5)]
print('Muy bajas o muy altas:', extremos)

print('Los que NO aprobaron:', notas[~(notas >= 3.0)])
````

### 5.2 Los parentesis no son opcionales

```python
notas[(notas >= 3.0) & (notas <= 4.5)]     # correcto
notas[notas >= 3.0 & notas <= 4.5]         # revienta
```

````python
print(notas[notas >= 3.0 & notas <= 4.5])

print('Con parentesis:', notas[(notas >= 3.0) & (notas <= 4.5)])

notas_curvadas = notas.copy()
notas_curvadas[notas_curvadas < 3.0] = 3.0
print('Originales:', notas)
print('Curvadas  :', notas_curvadas)
````

## Parte 6 - Dos dimensiones: tablas de numeros

`shape` ahora tiene dos numeros: `(3, 4)` = 3 filas y 4 columnas.
Para llegar a un dato se dan las dos coordenadas: `tabla[fila, columna]`.

````python
tabla = np.array([
    [4.5, 3.2, 4.0, 3.8],
    [2.8, 3.5, 4.2, 2.9],
    [5.0, 4.8, 4.9, 4.7]
])

print(tabla)
print('Forma (filas, columnas):', tabla.shape)
print('Dimensiones:', tabla.ndim)

print('Estudiante 1 en materia 2:', tabla[1, 2])

print('Todas las notas del estudiante 0:', tabla[0, :])

print('Todas las notas de la materia 3:', tabla[:, 3])
````

### 6.1 Pedir una fila que no existe

````python
print('La tabla tiene', tabla.shape[0], 'filas: las validas son 0, 1 y 2')

print('Estudiante 3:', tabla[3, 0])

print('Ultimo estudiante:', tabla[-1, :])
````

### 6.2 axis

| Escribes | Que hace |
|---|---|
| `tabla.mean()` | promedio de toda la tabla |
| `tabla.mean(axis=0)` | una cifra por columna (por materia) |
| `tabla.mean(axis=1)` | una cifra por fila (por estudiante) |

Regla: **`axis` es la dimension que desaparece.**

````python
print('Forma de la tabla:', tabla.shape, '-> 3 filas, 4 columnas')
print()

print('Promedio general      :', tabla.mean().round(2))

promedio_materia = tabla.mean(axis=0)
print('Promedio por materia  :', promedio_materia.round(2), '  <- salieron', promedio_materia.size)

promedio_estudiante = tabla.mean(axis=1)
print('Promedio por estudiante:', promedio_estudiante.round(2), '  <- salieron', promedio_estudiante.size)

print()
print('Nota maxima de cada materia:', tabla.max(axis=0))
print('Nota minima de cada estudiante:', tabla.min(axis=1))
````

### 7.1 Del diccionario de la Sesion 1 a tres listas paralelas

````python
supermercado = {
    'Frutas':    [('Manzana', 4500), ('Banano', 2800)],
    'Lacteos':   [('Leche', 4800), ('Queso', 12000), ('Yogur', 3200)],
    'Vegetales': [('Tomate', 3900), ('Cebolla', 2500)]
}

categorias = []
productos = []
precios = []

for categoria, lista_productos in supermercado.items():
    for nombre_producto, precio in lista_productos:
        categorias.append(categoria)
        productos.append(nombre_producto)
        precios.append(precio)

print('Categorias:', categorias)
print('Productos :', productos)
print('Precios   :', precios)
print('Las tres listas miden lo mismo:', len(categorias), len(productos), len(precios))
````

### 7.1b El descuento y la clasificacion, ahora vectorizados

`np.where(condicion, valor_si, valor_no)` es el `if / else` aplicado a todo el array.

````python
precios_np = np.array(precios)
categorias_np = np.array(categorias)

es_lacteo = categorias_np == 'Lacteos'
print('Mascara de lacteos:', es_lacteo)

precios_finales = np.where(es_lacteo, precios_np * 0.9, precios_np)
print('Precios originales:', precios_np)
print('Precios finales   :', precios_finales)

UMBRAL = 4000
etiquetas = np.where(precios_finales > UMBRAL, 'caro', 'barato')
print('Etiquetas         :', etiquetas)
````

### 7.2 pandas: el DataFrame (punto g de la actividad)

Un **DataFrame** es una tabla con nombres de columna. Se crea desde un diccionario:
cada clave es una columna, cada valor es la lista de datos de esa columna.

| Instruccion | Que hace |
|---|---|
| `df.head()` | muestra las primeras 5 filas |
| `df.shape` | (filas, columnas) |
| `df['Columna']` | saca una columna completa |

````python
import pandas as pd

df = pd.DataFrame({
    'Categoria': categorias,
    'Producto': productos,
    'Precio': precios,
    'PrecioFinal': precios_finales,
    'Etiqueta': etiquetas
})

print('Forma de la tabla:', df.shape)
print()

df.head()
````

### 7.2b Anadir una columna calculada

````python
df['Ahorro'] = df['Precio'] - df['PrecioFinal']

df['ConIVA'] = (df['PrecioFinal'] * 1.19).round(0)

print('Columnas ahora:', list(df.columns))
print()

print('Precio promedio:', df['PrecioFinal'].mean().round(0))
print('Producto mas caro:', df['Producto'][df['PrecioFinal'].idxmax()])
print()

promedio_por_categoria = df.groupby('Categoria')['PrecioFinal'].mean()
print('Promedio por categoria:')
print(promedio_por_categoria)

df.head()
````

### 7.3 matplotlib: el grafico de barras (punto h de la actividad)

Receta minima:
1. `plt.figure(figsize=(ancho, alto))`
2. `plt.bar(etiquetas, valores)`
3. `plt.title()`, `plt.xlabel()`, `plt.ylabel()`
4. `plt.show()`

````python
import matplotlib.pyplot as plt

plt.figure(figsize=(7, 4))
plt.bar(promedio_por_categoria.index, promedio_por_categoria.values, color='steelblue')
plt.title('Precio promedio por categoria')
plt.xlabel('Categoria')
plt.ylabel('Precio promedio (COP)')
plt.grid(axis='y', alpha=0.3)
plt.show()
````

### 7.3b La distribucion producto por producto

````python
colores_por_categoria = {'Frutas': 'seagreen', 'Lacteos': 'goldenrod', 'Vegetales': 'indianred'}

colores = [colores_por_categoria[c] for c in df['Categoria']]

plt.figure(figsize=(7, 4))
plt.barh(df['Producto'], df['PrecioFinal'], color=colores)
plt.title('Precio final por producto (color = categoria)')
plt.xlabel('Precio final (COP)')
plt.ylabel('Producto')
plt.grid(axis='x', alpha=0.3)
plt.tight_layout()
plt.show()

print('Con esto ya tienes los puntos g) y h) de la Actividad 1.')
````

## Resumen

| Concepto | Ejemplo | Para que |
|---|---|---|
| Importar | `import numpy as np` | traer la caja de herramientas |

| if vectorizado | `np.where(cond, si, no)` | decidir sobre toda la columna |
| Tabla 2D | `np.array([[..],[..]])` | filas y columnas |
| Direccion | `axis=0` columnas, `axis=1` filas | el eje que desaparece |
| DataFrame | `pd.DataFrame({...})` | tabla con nombres de columna |
