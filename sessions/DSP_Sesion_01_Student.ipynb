## 3. Variables: ponerle nombre a los datos

Una **variable** permite almacenar un valor para reutilizarlo posteriormente.

```text
nombre ─────► "Laura"
edad ───────► 31
compra ─────► 520000.0
activo ─────► True
```

````python
nombre = "Laura"
edad = 31
compra = 520000.0
activo = True

print(nombre)
print(edad)
print(compra)
print(activo)
````

Aunque todas son variables, sus valores representan cosas diferentes. Python reconoce esa diferencia mediante **tipos de datos**.

````python
print(type(nombre))
print(type(edad))
print(type(compra))
print(type(activo))
````

## 4. Tipos de datos fundamentales

| Tipo | Nombre en Python | Ejemplo |
|---|---|---|
| Texto | `str` | `"Tunja"` |
| Entero | `int` | `31` |
| Decimal | `float` | `520000.0` |
| Booleano | `bool` | `True`, `False` |
| Ausencia de valor | `NoneType` | `None` |

Un mismo símbolo puede representar cosas muy diferentes según su tipo.

````python
a = 31
b = "31"

print(a, type(a))
print(b, type(b))
````

## 9. Decisiones con `if`

```text

                │            │
                ▼            ▼

```

````python
compra = 1_250_000

if compra > 1_000_000:
    print("Compra alta")
````

### `if`, `elif`, `else`

Podemos representar más de dos escenarios.

````python
compra = 780_000

if compra < 500_000:
    print("Compra baja")
elif compra <= 1_000_000:
    print("Compra media")
else:
    print("Compra alta")
````

## 10. Funciones: convertir una solución en algo reutilizable

Una función encapsula una tarea.

```text
ENTRADA
  │
  ▼
┌─────────────────────┐
│ clasificar_compra() │
└─────────────────────┘
  │
  ▼
SALIDA
```

````python
def clasificar_compra(valor):
    if valor < 500_000:
        return "Baja"
    elif valor <= 1_000_000:
        return "Media"
    else:
        return "Alta"

print(clasificar_compra(250_000))
print(clasificar_compra(850_000))
print(clasificar_compra(1_500_000))
````

### ¿Por qué usar funciones?

```text
Sin función:
dato 1 → escribir lógica
dato 2 → repetir lógica
dato 3 → repetir lógica

Con función:
dato 1 ─┐
dato 2 ─┼──► clasificar_compra() ───► resultado
dato 3 ─┘
```
